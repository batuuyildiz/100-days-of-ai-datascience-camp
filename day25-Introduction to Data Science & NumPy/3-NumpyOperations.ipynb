```python
data = np.array([10,20,12,13,25,100,105])
```

```python
# outlier
```

```python
mean=np.mean(data)
```

```python
mean
```

```python
std=np.std(data)
```

```python
std
```

```python
z_score=(data-mean)/std
```

```python
z_score
```

```python
z_score[z_score>1]
```
