# Sets

```python
# unique elements, unordered
```

```python
myList = [10, 20, 30, 40, 50, 10,10,20,30,40,50]
```

```python
len(myList)  # Length of the list
```

```python
mySet=set(myList)  # Convert to a set to remove duplicates
```

```python
mySet
```

```python
mySet={10, 20, 30, 40, 50,10,10,20,30,40,50}  # This will still only keep unique elements   
```

```python
mySet
```

```python
# mySet.[0]  # This will not work, sets do not support indexing
```

```python
mySet = {10, 20, 30, 40, 50}
mySet2 = {30, 40, 50, 60, 70}
```

```python
mySet.union(mySet2)  # Union of two sets
```

```python
mySet.intersection(mySet2)  # Intersection of two sets
```

```python
countriesList = ["de", "fr", "it", "es", "de", "fr", "it", "es","tr", "tr", "tr"]
```

```python
len(countriesList)  # Length of the list
```
