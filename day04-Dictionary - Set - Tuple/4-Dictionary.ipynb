```python
fitness_dictionary = {"banana": 105, "apple": 95, "orange": 62, "kiwi": 42}
```

```python
type(fitness_dictionary)
```

```python
# fitness_dictionary[0]  # Accessing the first element of the dictionary will raise an error because dictionaries are not indexed by position
```

```python
fitness_dictionary["banana"]  # Accessing the value for the key "banana"
```

```python
fitness_dictionary.keys()  # Returns a list of keys in the dictionary
```

```python
fitness_dictionary.values()  # Returns a list of values in the dictionary
```

```python
type(fitness_dictionary.values())  # Returns the type of the values in the dictionary
```

```python
list(fitness_dictionary.values())  # Converts the values to a list
```

```python
fitness_dictionary["banana"] = 110  # Updating the value for the key "banana"
```

```python
fitness_dictionary
```

```python
fitness_dictionary["melon"] = 50  # Adding a new key-value pair to the dictionary
```

```python
fitness_dictionary
```
