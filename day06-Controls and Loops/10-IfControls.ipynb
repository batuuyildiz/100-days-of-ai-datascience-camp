```python
if x < y:
    print("x is less than y")
elif x > y:
    print("x is greater than y")
else:
    print("x is equal to y")    
```

```python
my_superhero = "Superman"
if my_superhero == "Superman":  
    print("My superhero is Superman")
elif my_superhero == "Batman":
    print("My superhero is Batman")
else:
    print("My superhero is not Superman or Batman")
```

```python
user_superhero = input("Enter your favorite superhero: ")
if user_superhero == "Superman":
    print("Your favorite superhero is Superman")
elif user_superhero == "Batman":
    print("Your favorite superhero is Batman")
else:
    print("Your favorite superhero is "+ user_superhero)
```

```python
my_superhero_list = ["Superman", "Batman", "Wonder", "Flash"]
if "Superman" in my_superhero_list:
    print("Superman is in the list")
```
