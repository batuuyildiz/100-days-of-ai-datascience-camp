```python
class Person():
    #properties, attributes
    name=""
    age=0
    gender=""
    #method, initializer
    def __init__(self,nameInput,ageInput,genderInput):# self is the instance of the class
        self.name=nameInput
        self.age=ageInput   
        self.gender=genderInput
```

```python
batuhan=Person("Batuhan",30,"Male")
```

```python
class Dog():

    year = 7
    def __init__(self,age):
        self.age=age
    def humanAge(self):
        return self.age * self.year #Dog.year->self.year
```
