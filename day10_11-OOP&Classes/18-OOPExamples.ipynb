# Inheritance

```python
class Musician():

    def __init__(self,name):
        self.name=name
        print("Musician class")
    def test1(self):
        print("Test1 from Musician class")
    def test2(self):
        print("Test2 from Musician class")
```

```python
batu=Musician("Batu")
```

```python
batu.name
```

```python
batu.test1()
```

```python
batu.test2()
```

```python
class MusicianPlus(Musician):# Inheritance
    def __init__(self, name):
        Musician.__init__(self,name) # Call the constructor of the parent class
        print("MusicianPlus class")
```

```python
atlas = MusicianPlus("Atlas")
```

```python
atlas.test1()
atlas.test2()
```

```python
atlas.name="Batuhan Yildiz"
```

```python
type(atlas)
```

# Polymorphism

```python
class Banana():

    def __init__(self,name):
        self.name=name
        
    def info(self):
        return f"100 calories {self.name}"
class Apple():

    def __init__(self,name):
        self.name=name
        
    def info(self):
        return f"50 calories {self.name}"
```

```python
banana=Banana("banana")
```

```python
apple=Apple("apple")
```

```python
banana.info()
```

```python
apple.info()
```

```python
fruitList=[banana,apple]
```

```python
for fruit in fruitList:
    print(fruit.info())
```

# Encapsulation

```python
class Phone():
    def __init__(self,name,price):
        self.name = name
        self.__price = price

    def info(self):
        print(f"{self.name} price is: {self.__price}") # Private attribute  

    def changePrice(self,price):
        self.__price = price
```

```python
iphone = Phone("Iphone 14", 30000)
```

```python
iphone.info()
```

```python
iphone.__price=400 # This will not change the private attribute
```

```python
iphone.info()   # The price is still 30000
```

```python
iphone.changePrice(400)
```

```python
iphone.info()
```

# Abstraction

```python
# Abstraction
```

```python
from abc import ABC, abstractmethod # Abstract Base Class
```

```python
class Car (ABC):
    @abstractmethod
    def maxSpeed():
        pass
```

```python
# myCar=car() # This will give an error because we cannot instantiate an abstract class
```

```python
class Tesla(Car):

    def maxSpeed(self):
        print("200 km/h")
    
```

```python
tesla=Tesla()
```

```python
tesla.maxSpeed()
```

```python
class Mercedes(Car):

    def maxSpeed(self):
        print("250 km/h")
```

# Special Method

```python
class fruit():
    def __init__(self,name,calories):
        self.name=name
        self.calories=calories
    def __str__(self):
        return f"{self.name} has {self.calories} calories"
    def __len__(self):
        return self.calories
```

```python
myFruit=fruit("banana",150)
```

```python
myFruit.calories
```

```python
print(myFruit)
```

```python
class Train():

    def __init__(self,name):
        self.name=name
    def __getitem__(self,key):
        if key=="a":
            return self.name
        else:
            return "Not found"
```

```python
myTrain=Train("Express")
```

```python
myTrain["a"]
```
