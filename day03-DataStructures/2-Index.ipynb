# slicing, starting index, stopping index, steptipng size

```python
barcode[::]
```

```python
barcode[3::] # start from index 3 to the end     
```

```python
barcode[:3:]# start from the beginning to index 3
```

```python
barcode[:5:1] # start from the beginning to index 5 with step size 1
```

```python
barcode[3:9:2]  
```

```python
barcode[::-1]   # reverse the string   
```
